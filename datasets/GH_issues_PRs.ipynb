Source paper: PI-Link: A Ground-Truth Dataset of Links Between Pull-Requests and Issues in GitHub

paper link: https://ieeexplore.ieee.org/abstract/document/10002372

According to the description it should consist of 6 tables: Issue, Repository, Commit, Label, Comment, PullRequest. However, I couldn't locate the majority of those tables on kaggle or in other linked sources. The data published contains issues table and table with relations between issue_url and pr_url. Therefore I'm not sure if it will be sufficient for our needs.

```python
import kagglehub
import pandas as pd
import os
```

The code to download the dataset:

```python
path = kagglehub.dataset_download("zakareaalshara/dataset")

print("Path to dataset files:", path)

csv_files = [f for f in os.listdir(path) if f.endswith(".csv")]
```

Merging the tables, so that pull request url is in the same table as other issue data:

```python
df0 = pd.read_csv(os.path.join(path, csv_files[0]))
df1 = pd.read_csv(os.path.join(path, csv_files[1]))
print(f"Loaded: {csv_files[0]} {df0.shape}, {csv_files[1]} {df1.shape}")

merged_table = pd.merge(df0, df1, left_on="url", right_on="issue_url", how="inner")
print(f"\nMerged shape: {merged_table.shape}")
print(f"\nColumns: {list(merged_table.columns)}")
```
